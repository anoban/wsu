```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
summary = pd.read_csv(r"./994species_5states_model_averages.csv")
summary
```

```python
summary.query(r"trait=='logRD'")
```

```python
COLUMNS_OF_INTEREST = ["rates", "alpha", "sigma.sq", "theta"]
```

```python
COLS = {
    "AM": "red", "NM": "blue", "NMAM": "purple", "EcMAM": "orange", "EcM": "yellow"
}
```

```python
plt.rcParams["font.size"] = 20
fig, axes = plt.subplots(nrows=2, ncols=4, sharex=True, sharey=False)
fig.set_size_inches(40, 20)

for (axis, col) in zip(axes[0], COLUMNS_OF_INTEREST): # first row
    axis.scatter(x=summary.query(r"trait=='logRD' and is_cd==True").tip_state,
                 y=summary.query(r"trait=='logRD' and is_cd==True").loc[:, col], s=150, edgecolor = "black",
                 color="red"
                )
    axis.set_title(col.title())
    axis.grid()
    axis.set_axisbelow(True)

for (axis, col) in zip(axes[1], COLUMNS_OF_INTEREST): # first row
    axis.scatter(x=summary.query(r"trait=='logRD' and is_cd==False").tip_state,
                 y=summary.query(r"trait=='logRD' and is_cd==False").loc[:, col], s=150, edgecolor = "black",
                 color="red"
                )
    axis.set_title(col.title())
    axis.grid()
    axis.set_axisbelow(True)
    
plt.tight_layout()
# plt.show()
plt.savefig(fname=r"../../plots/houwie_5states_994sp_summary_logrd.png", dpi=800)
```
