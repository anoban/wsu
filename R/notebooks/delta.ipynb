```r
library("ape")
source("../thirdparty/delta.R")
```

```r
data <- read.csv("../../data/chapter2/FRED/subsets/name_matched_FRED4_1301.csv")
phylogeny <- ape::read.tree("../../data/chapter2/uphylomaker/FRED4_1301.tre")

# make sure there's no zero length branches
if(sum(phylogeny$edge.length == 0)) phylogeny$edge.length[phylogeny$edge.length == 0] <- quantile(phylogeny$edge.length, 0.1) * 0.1 # replace the 0 length branches with 0.1 * 10th percentile branch length
stopifnot(all(phylogeny$edge.length > 0))
if(!ape::is.binary(phylogeny)) phylogeny <- ape::multi2di(phylogeny)
```

```r
ape::is.rooted(phylogeny)
```

```r
ape::is.binary(phylogeny)
```

```r
delta(trait = setNames(data$state, nm = data$binominal), tree = phylogeny, lambda0 = 0.1, se = 0.0589, sim = 10000, thin = 10, burn = 100)
```
