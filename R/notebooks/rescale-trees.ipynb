# ___Rescaling Phylogenies___
-----------------------

```r
library("geiger")
library("ape")
library("phytools")
```

```r
phylogeny <- ape::read.tree("../../data/chapter2/uphylomaker/FRED4_1301.tre")
if(!ape::is.binary(phylogeny)) phylogeny <- ape::multi2di(phylogeny)
```

```r
# all the parameters are random values

lambda_tree <- phytools::rescale(phylogeny, model = "lambda", lambda = 0.96)
ou_tree <- phytools::rescale(phylogeny, model = "OU", alpha = 0.00009, sigsq = 0.9500)
white_noise_tree <- phytools::rescale(phylogeny, model = "white", sigsq = 0.9999)
eb_tree <- phytools::rescale(phylogeny, model = "EB", a = 1.000, sigsq = 0.5)
kappa_tree <- phytools::rescale(phylogeny, model = "kappa", kappa = 0.7800, sigsq = 0.5)
bm_tree <- phytools::rescale(phylogeny, model = "BM", sigsq = 0.5)
```

```r
# all the parameters have been tuned to match the original phylogeny

options(repr.plot.width = 15, repr.plot.height = 25, repr.plot.res = 150) 
par(mfcol = c(1, 7), mar = c(1, 1, 1, 1))

plot(phylogeny, show.tip.label = FALSE)
plot(lambda_tree, show.tip.label = FALSE)
plot(ou_tree, show.tip.label = FALSE)
plot(white_noise_tree, show.tip.label = FALSE)
plot(eb_tree, show.tip.label = FALSE)
plot(kappa_tree, show.tip.label = FALSE)
plot(bm_tree, show.tip.label = FALSE)
```
