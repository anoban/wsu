# ___Rescaling Phylogenies___
-----------------------

```r
library("geiger")
library("ape")
library("phytools")
```

```r
phylogeny <- ape::read.tree("../../data/chapter2/uphylomaker/FRED4_1301.tre")
if(!ape::is.binary(phylogeny)) phylogeny <- ape::multi2di(phylogeny)
```

```r
lambda_tree <- phytools::rescale(phylogeny, model = "lambda", lambda = 0.56)
ou_tree <- phytools::rescale(phylogeny, model = "OU", alpha = 0.999, sigsq = 1.000)
```

```r
options(repr.plot.width = 15, repr.plot.height = 16, repr.plot.res = 150)
par(mfcol = c(1, 3))
plot(phylogeny)
plot(lambda_tree)
plot(ou_tree)
```
