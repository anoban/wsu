```python
import re
import os
from urllib.request import urlopen
from bs4 import BeautifulSoup
```

```python
with urlopen("http://www.phytools.org/Rbook") as req:
    resp = req.read()
```

```python
soup = BeautifulSoup(resp, "html.parser")
```

```python
corrections_n_supplements = soup.find(name="div", attrs={"id": "corrections-data-files-and-other-supplementary-information"})
```

```python
corrections_n_supplements
```

```python
# the above div is divided into chapters
corrections_n_supplements.find_all(name="h3")
```

```python
corrections_n_supplements.find_all(name="ol")
```

```python
# collect all the download links
# every link is in the format chapterno/filename

files = [li.find(name='a')["href"] for ol in corrections_n_supplements.find_all(name="ol") for li in ol.find_all(name="li")]
```

```python
DIRNO = re.compile(r"\d+") 
```

```python
for dirno in set([DIRNO.findall(p)[0] for p in files]): # extract the chapter numbers and pre create dirs for downloading the supplements
    os.mkdir(path=f"../PCMiR/{dirno}")
```

```python
for file in files:
    with urlopen(f"http://www.phytools.org/Rbook/{file}") as req:
        with open(f"../PCMiR/{file}", mode="wt") as fp:
            fp.write(req.read().decode(encoding="latin")) # utf8 causes decoding errors
```
