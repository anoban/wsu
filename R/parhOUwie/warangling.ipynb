# ___Name matching phylogenies and trait data before the parallel fits of `hOUwie`___
-------------------------------

```r
library("ape")
```

```r
rd1005sp <- read.csv("./genus_state_rec_logged_species_avgd_RD_1005sp.csv")
srl1005sp <- read.csv("./genus_state_rec_logged_species_avgd_SRL_1005sp.csv")
tree1005sp <- ape::read.tree("./1005sp.tre")
```

```r
all(rd1005sp$binominal[match(tree1005sp$tip.label, rd1005sp$binominal)] == tree1005sp$tip.label)
all(srl1005sp$binominal[match(tree1005sp$tip.label, srl1005sp$binominal)] == tree1005sp$tip.label)
```

```r
rd1005sp <- rd1005sp[match(tree1005sp$tip.label, rd1005sp$binominal), ]
srl1005sp <- srl1005sp[match(tree1005sp$tip.label, srl1005sp$binominal), ]
```

```r
all(rd1005sp$binominal == tree1005sp$tip.label)
all(srl1005sp$binominal == tree1005sp$tip.label)
```

```r
write.csv(rd1005sp, "./genus_state_rec_logged_species_avgd_RD_1005sp.csv", row.names=FALSE)
write.csv(srl1005sp, "./genus_state_rec_logged_species_avgd_SRL_1005sp.csv", row.names=FALSE)
```

```r
# 994 species subsets

tree994 <- ape::read.tree("./4states_994sp.tre")
rd994_4states <- read.csv("./gen_rec_4state_logged_994species_avgd_RD.csv")
srl994_4states <- read.csv("./gen_rec_4state_logged_994species_avgd_SRL.csv")
rd994_5states <- read.csv("./gen_rec_5state_logged_994species_avgd_RD.csv")
srl994_5states <- read.csv("./gen_rec_5state_logged_994species_avgd_SRL.csv")
```

```r
all(rd994_4states$binominal==rd994_5states$binominal)
all(srl994_5states$binominal==srl994_4states$binominal)
all(rd994_4states$binominal==srl994_4states$binominal)

# cool, we can then just use one reordered index
```

```r
all(rd994_4states$binominal[match(tree994$tip.label, rd994_4states$binominal)] == tree994$tip.label)
```

```r
matched_idx <- match(tree994$tip.label, rd994_4states$binominal)
```

```r
rd994_4states <- rd994_4states[matched_idx, ]
srl994_4states <-srl994_4states[matched_idx, ]
rd994_5states <- rd994_5states[matched_idx, ]
srl994_5states <-srl994_5states[matched_idx, ]
```

```r
stopifnot(all(tree994$tip.label==rd994_4states$binominal))
stopifnot(all(tree994$tip.label==rd994_5states$binominal))
stopifnot(all(tree994$tip.label==srl994_4states$binominal))
stopifnot(all(tree994$tip.label==srl994_5states$binominal))
```

```r
write.csv(rd994_4states, "./gen_rec_4state_logged_994species_avgd_RD.csv", row.names = FALSE)
write.csv(srl994_4states, "./gen_rec_4state_logged_994species_avgd_SRL.csv", row.names = FALSE)
write.csv(rd994_5states, "./gen_rec_5state_logged_994species_avgd_RD.csv", row.names = FALSE)
write.csv(srl994_5states, "./gen_rec_5state_logged_994species_avgd_SRL.csv", row.names = FALSE)
```
