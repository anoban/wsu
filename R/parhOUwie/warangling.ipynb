```r
library("ape")
```

```r
rddata <- read.csv("./genus_state_rec_logged_species_avgd_RD_1005sp.csv")
srldata <- read.csv("./genus_state_rec_logged_species_avgd_SRL_1005sp.csv")
tree <- ape::read.tree("./FRED_subset_collab_1005sp.tre")
```

```r
all(rddata$binominal[match(tree$tip.label, rddata$binominal)] == tree$tip.label)
```

```r
all(srldata$binominal[match(tree$tip.label, srldata$binominal)] == tree$tip.label)
```

```r
rddata <- rddata[match(tree$tip.label, rddata$binominal), ]
srldata <- srldata[match(tree$tip.label, srldata$binominal), ]
```

```r
all(rddata$binominal == tree$tip.label)
```

```r
all(srldata$binominal == tree$tip.label)
```

```r
write.csv(rddata, "./genus_state_rec_logged_species_avgd_RD_1005sp.csv", row.names=FALSE)
```
