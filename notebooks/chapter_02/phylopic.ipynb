```python
from urllib.request import Request, urlopen
import pandas as pd
import json
from typing import NamedTuple
```

```python
# API docs can be found at http://api-docs.phylopic.org/v2/
# https://www.phylopic.org/articles/api-recipes
```

```python
with urlopen(Request(url=r"https://api.phylopic.org/images")) as req:
    resp = req.read()
```

```python
resp.decode() # total 268 pages with 12863 items
```

```python
all_pages = []

for i in range(268):
    with urlopen(f"https://api.phylopic.org/images?build=545&page={i}") as req:
        all_pages.append(req.read().decode())
```

```python
# put together a clean lookup table with species names and urls
```

```python
len(all_pages)
```

```python
json.loads(all_pages[0])["_links"]
```

```python
json.loads(all_pages[0])["_links"]["self"]
```

```python
json.loads(all_pages[0])["_links"]["items"]
```

```python
class PhyloPicRecord(NamedTuple):
    binominal: str
    url: str
```

```python
[PhyloPicRecord(_["title"], f"https://www.phylopic.org{_["href"]}") for _ in json.loads(all_pages[0])["_links"]["items"]]
```

```python
phylopic_parsed = [PhyloPicRecord(_["title"], f"https://www.phylopic.org{_["href"]}") for page in all_pages for _ in json.loads(page)["_links"]["items"]]
```

```python
len(phylopic_parsed) # got everything :)
```

```python
phylopic_full_catelog = pd.DataFrame(phylopic_parsed).sort_values(by="binominal").reset_index(drop=True)
phylopic_full_catelog
```

```python
phylopic_full_catelog.to_csv(r"../../data/chapter2/PhyloPic_build545_full_catelog.csv", index=False)
```
