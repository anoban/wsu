```python
from PIL import Image

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from skimage.filters import median, gaussian
from skimage.color import rgb2gray
from skimage.util import invert
from skimage.morphology import skeletonize, thin, medial_axis
```

```python
root = np.array(Image.open(r"../../2025_10_08_11_37_58_002.png").convert("RGB"))
greyscale = rgb2gray(root)
inverted = invert(greyscale)
```

```python
skeleton = skeletonize(inverted).astype(np.uint8)
medial = medial_axis(inverted)
```

```python
fig, axes = plt.subplots()
axes.imshow(inverted, cmap="grey")
axes.set_axis_off()
plt.tight_layout()
# plt.show()
plt.savefig(r"../../plots/invgrey.png", format=r"png", dpi=900)
```

```python
fig, axes = plt.subplots()
axes.imshow(medial, cmap="grey")
plt.show()
```

```python
fig, axes = plt.subplots()
axes.imshow(median(inverted, mode="constant"), cmap="grey")
plt.show()
```

```python
fig, axes = plt.subplots()
axes.imshow(gaussian(inverted, sigma=1.995), cmap="grey")
plt.show()
```
