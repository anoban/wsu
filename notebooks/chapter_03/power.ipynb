```python
import sys
from itertools import permutations, combinations
sys.path.append(r"../lib/")

import numpy as np
np.random.seed(2025 - 10 - 31)
from numba import jit, njit
from scipy.stats import pearsonr
import matplotlib.pyplot as plt
```

```python
experiment = np.eye(N=6)
```

```python
fig, axes = plt.subplots()
fig.set_size_inches(5, 5)
# axes.set_axis_off()
axes.imshow(experiment, cmap="YlGn")
axes.hlines(y=np.linspace(start=axes.get_ylim()[0], stop=axes.get_ylim()[1], num=7), xmin=axes.get_xlim()[0], xmax=axes.get_xlim()[1], colors="black")
axes.vlines(x=np.linspace(start=axes.get_xlim()[0], stop=axes.get_xlim()[1], num=7), ymin=axes.get_ylim()[0], ymax=axes.get_ylim()[1], colors="black")
axes.set_xticks(ticks=np.linspace(start=0, stop=5, num=6), labels=["TT4", "TT6", "TT7", "TT13", "TT15", "TT17"])
axes.set_yticks(ticks=np.linspace(start=0, stop=5, num=6), labels=[_.replace("TT", "AMF") for _ in ("TT4", "TT6", "TT7", "TT13", "TT15", "TT17")])
plt.tight_layout()
plt.show()
# plt.savefig(r"../../plots/expdesignch3.png", format=r"png", dpi=700)
```

```python
TICKS = np.linspace(start=0, stop=5, num=6).astype(np.int32)

fig, axes = plt.subplots()
fig.set_size_inches(5, 5)
for (x, y) in zip(*np.where(experiment != 1)): experiment[x, y] = np.random.rand() / 1.50
axes.imshow(experiment, cmap="Reds")

[axes.text(x=x-.2, y=y+.09, s=f"{experiment[x, y]:.2f}", color="white") for x in range(6) for y in range(6)]

axes.hlines(y=np.linspace(start=axes.get_ylim()[0], stop=axes.get_ylim()[1], num=7), xmin=axes.get_xlim()[0], xmax=axes.get_xlim()[1], colors="black")
axes.vlines(x=np.linspace(start=axes.get_xlim()[0], stop=axes.get_xlim()[1], num=7), ymin=axes.get_ylim()[0], ymax=axes.get_ylim()[1], colors="black")
axes.set_xticks(ticks=TICKS, labels=["TT4", "TT6", "TT7", "TT13", "TT15", "TT17"])
axes.set_yticks(ticks=TICKS, labels=[_.replace("TT", "AMF") for _ in ("TT4", "TT6", "TT7", "TT13", "TT15", "TT17")])
plt.tight_layout()
plt.show()
# plt.savefig(r"../../plots/expdesignch3_power.png", format=r"png", dpi=700)
```
