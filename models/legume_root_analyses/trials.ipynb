```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from skimage.color import rgb2gray # Y = 0.2125 R + 0.7154 G + 0.0721 B
```

```python
root = np.array(Image.open(r"../../root.jpeg"))
root.shape # height, width, colour channels
```

```python
greyscale = rgb2gray(root)
```

```python
fig, axes = plt.subplots(ncols=2)
fig.set_size_inches(11, 7.5)

for ax, im in zip(axes, [root, greyscale]):
    ax.imshow(im, cmap="binary")
    ax.set_axis_off()
plt.show()
```

```python
# [B, G, R]
(root * np.array([0.0721, 0.7154, 0.2125], dtype=np.float32)).sum(axis=2) / 255.0
```
